```python
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors= 'coerce')
df.isnull().sum()
```

```python
# Rows where TotalCharges is missing
missing_total = df[df['TotalCharges'].isna()]
print(f"Missing TotalCharges rows: {len(missing_total)}")

# Show key columns for inspection
cols_to_show = ['customerID', 'tenure', 'MonthlyCharges', 'TotalCharges', 'Churn']
print(missing_total[cols_to_show].head(20))
```

```python
## BEFORE FILLNA CHANGE
## ------
no_tenure = df[df['tenure'] == 0]
if missing_total.equals(no_tenure):
    print("All missing TotalCharges correspond to tenure of 0.")
print(no_tenure[cols_to_show].head(20))
## ------

print("\n=== NOW RUNNING FILLNA ===")
# Since these customers have no tenure, we can fill TotalCharges with 0
df['TotalCharges'] = df['TotalCharges'].fillna(0)
print(f"Missing TotalCharges after fill: {df['TotalCharges'].isna().sum()}")


## AFTER FILLNA CHANGE
## -----
no_tenure_updated = df[df['tenure'] == 0]
print("Updated no_tenure rows after fillna:")
no_tenure_updated[cols_to_show]
## ------


# no_tenure is snapshot of original df with total charges missing
# no_tenure_updated is snapshot of orignal df AFTER total charges fillna
```

```python
churn_count = df['Churn'].value_counts()
churn_percentage = df['Churn'].value_counts(normalize=True) *100
churn_df = pd.DataFrame({'Count': churn_count, 'Percentage': churn_percentage})
churn_df
```

```python
df['SeniorCitizen'] = df['SeniorCitizen'].map({1: 'Yes', 0: 'No'})
df['SeniorCitizen']
```

```python
df['Churn'] = df['Churn'].map({'Yes': 1, 'No': 0})
numeric_features = df.select_dtypes(include=[np.number]).columns.tolist()
numeric_features.remove('Churn')

categorical_features = df.select_dtypes(exclude=[np.number]).columns.tolist()
categorical_features.remove('customerID')
print("Numerical Features:", numeric_features)
print("Categorical Features:", categorical_features)
for col in categorical_features:
    uniques = df[col].unique()
    print(f"\nColumn: {col}")
    print(f"Unique count: {len(uniques)}")
    vc = df[col].value_counts(dropna=False)
    print(vc.head(50))
for col in numeric_features:
    print(f"\nColumn: {col}")
    print(df[col].describe())
```

IMPORTANT TO NOTE: 

SeniorCitizen
Originally: Num (0 and 1)
Changed: Cat (No and Yes)

Churn 
Originally: Cat (No and Yes)
Changed: Num (0 and 1)

TotalCharges
Originally: Cat (Missing Values)
Changed: Numeric (NaN)

```python
Q1 = df[numeric_features].quantile(0.25)
Q3 = df[numeric_features].quantile(0.75)
IQR = Q3 - Q1

outliers = ((df[numeric_features] < (Q1 - 1.5 * IQR)) | 
            (df[numeric_features] > (Q3 + 1.5 * IQR)))
print("Outlier counts per feature:")
print(outliers.sum())
```
