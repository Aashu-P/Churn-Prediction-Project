```python
import joblib
import pandas as pd
import numpy as np
preprocessor = joblib.load('../models/preprocessor.pkl')
model = joblib.load('../models/xgb_model.pkl')
threshold = joblib.load('../models/threshold.pkl')
```

```python
#Simulating 2 user inputs (raw and messy with missing values)
user1_input = pd.DataFrame([{
    "gender": "Female",
    "SeniorCitizen": 0,
    "Partner": "Yes",
    "Dependents": None,
    "tenure": 5,
    "PhoneService": "Yes",
    "MultipleLines": None,
    "InternetService": "Fiber optic",
    "OnlineSecurity": "No",
    "OnlineBackup": "No",
    "DeviceProtection": "No",
    "TechSupport": "No",
    "StreamingTV": "Yes",
    "StreamingMovies": "Yes",
    "Contract": "Month-to-month",
    "PaperlessBilling": "Yes",
    "PaymentMethod": "Electronic check",
    "MonthlyCharges": 89.9,
    "TotalCharges": None
}])
user2_input = pd.DataFrame([{
    "gender": "Male",
    "SeniorCitizen": 1,
    "Partner": "Yes",
    "Dependents": None,
    "tenure": 3,
    "PhoneService": "Yes",
    "MultipleLines": "Yes",
    "InternetService": None,
    "OnlineSecurity": "No",
    "OnlineBackup": None,
    "DeviceProtection": "No",
    "TechSupport": "No",
    "StreamingTV": "Yes",
    "StreamingMovies": "No",
    "Contract": None,
    "PaperlessBilling": "Yes",
    "PaymentMethod": "Mailed check",
    "MonthlyCharges": 89.9,
    "TotalCharges": None
}])
user_inputs = pd.concat([user1_input, user2_input], ignore_index=True)
user_inputs
```

```python
user_preprocessed = preprocessor.transform(user_inputs)
churn_probability = model.predict_proba(user_preprocessed)[:, 1]
churn_prediction = np.where(churn_probability >= threshold, 'Yes', 'No')

#zip pairs prob and pred x
for i, (prob,pred) in enumerate(zip(churn_probability, churn_prediction), start=1):
    print(f"User {i}: Churn Probability = {prob*100:.2f}%, Predicted Churn = {pred}")
```
