```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(r"C:\Users\aashu\OneDrive\Desktop\Churn Prediction\data\raw_churn_data.csv")
```

```python
numeric_features = df.select_dtypes(include=[np.number]).columns.tolist()

categorical_features = df.select_dtypes(exclude=[np.number]).columns.tolist()
categorical_features.remove('customerID')
print("Numerical Features:", numeric_features)
print("Categorical Features:", categorical_features)
for col in categorical_features:
    uniques = df[col].unique()
    print(f"\nColumn: {col}")
    print(f"Unique count: {len(uniques)}")
    vc = df[col].value_counts(dropna=False)
    print(vc.head(50))
for col in numeric_features:
    print(f"\nColumn: {col}")
    print(df[col].describe())
```

```python
binary_cols = ['Churn', 'Partner', 'Dependents', 'PhoneService', 'PaperlessBilling']
ohe_cols = ['gender', 'MultipleLines', 'InternetService', 'OnlineSecurity', 'OnlineBackup',
            'DeviceProtection', 'TechSupport', 'StreamingTV', 'StreamingMovies', 'Contract', 'PaymentMethod']

df_encoded = df.copy()
df_encoded['TotalCharges'] = pd.to_numeric(df_encoded['TotalCharges'], errors='coerce')
df_encoded['TotalCharges_encoded'] = df_encoded['TotalCharges'].fillna(0)
df_encoded['Churn_encoded'] = df_encoded['Churn'].map({'Yes': 1, 'No': 0})
df_encoded['Partner_encoded'] = df_encoded['Partner'].map({'Yes': 1, 'No': 0})
df_encoded['Dependents_encoded'] = df_encoded['Dependents'].map({'Yes': 1, 'No': 0})
df_encoded['PhoneService_encoded'] = df_encoded['PhoneService'].map({'Yes': 1, 'No': 0})
df_encoded['PaperlessBilling_encoded'] = df_encoded['PaperlessBilling'].map({'Yes': 1, 'No': 0})

df_encoded = pd.get_dummies(df_encoded, columns=ohe_cols, drop_first=True, dtype=int)
df_encoded.drop(columns=binary_cols + ['TotalCharges', 'customerID'], inplace=True)
```
